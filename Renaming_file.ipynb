```python
#Code from vs-copilot
import os
import glob

for filename in glob.glob("Cropped_germination_pictures/*.png"):
    full_path = filename
    base_name = os.path.basename(filename)
    stem, ext = os.path.splitext(base_name)

    # Split only the filename stem, not the extension
    if "_" in stem:
        prefix, day_part = stem.rsplit("_", 1)

        # Example: day_part = "D1"
        if day_part.startswith("D") and day_part[1:].isdigit() and len(day_part[1:]) == 1:
            new_day = "D" + day_part[1:].zfill(2)
            new_stem = prefix + "_" + new_day
            new_name = new_stem + ext
            new_path = os.path.join(os.path.dirname(full_path), new_name)

            os.rename(full_path, new_path)
            print(f"Renamed: {base_name} -> {new_name}")
```
